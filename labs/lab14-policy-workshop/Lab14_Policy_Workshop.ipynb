## Part 1 — Why a firm needs an AI policy (3 min read)
- **Confidentiality.** Client drawings, reports, and fees usually belong to the client or are covered by contract.
  Pasting them into a public tool shares them with a third party, and some tools may keep or reuse what you type
  (it depends on the tool's settings and terms — check them).
- **Public safety and responsible charge.** The NSPE and ASCE codes of ethics ask engineers to hold public safety
  paramount and to work only within their competence. A licensed engineer in **responsible charge** (direct control
  and personal supervision of the work) remains responsible for sealed work **no matter what tools were used**.
- **Quality.** You saw it in Labs 08 and 12: fluent answers with wrong formulas, wrong units, and made-up citations.
- **Honesty with clients.** Clients may ask how AI is used on their project, and some contracts now say so.
- **Ownership.** Who owns AI-generated text and images is still being settled. In the U.S., the Copyright Office has
  said that material generated entirely by AI, without enough human creative contribution, is not protected by
  copyright (as of 2026 — check current guidance).
- **Rules change fast.** Laws, standards, and tool terms change often, so a good policy names an owner and a review
  cycle.

| Term | Plain meaning |
|---|---|
| **Data classification** | Sorting information by how much harm it would cause if it leaked. The course uses four classes: **Public / Internal / Confidential (client) / Restricted (personal & security-sensitive)**. |
| **Approved tool** | A tool the firm has reviewed and allowed, for stated kinds of data. |
| **Consumer vs enterprise tool** | A free/personal account vs a firm-licensed account with a contract about how data are handled. |
| **Responsible charge** | The licensed engineer who directs and supervises the work and is professionally responsible for it. |
| **Disclosure** | Telling the client (or others) that AI was used, and how. |
| **Incident** | Something that went wrong or nearly went wrong — e.g., client data in the wrong tool, an AI error in issued work. |

## Part 2 — Write the policy (20 min)
### 2.1 What may go into which tool? (whole team, 7 min)
The heart of most firm policies is a table: **which kind of information may go into which kind of tool.**

**Data classes** (the same four classes used everywhere in this course)
- **Public** — already public: OSHA rules, a city's published zoning code, the firm's website, published papers.
- **Internal** — the firm's own non-client information: standard-detail templates, training slides, draft marketing text.
- **Confidential (client)** — client and project information: drawings, calculations, geotechnical reports, fees and
  bids, RFIs (requests for information from contractors), project photos, contracts, meeting notes.
- **Restricted (personal & security-sensitive)** — the highest risk: personal data (employee records, recordings of
  people, occupant data, faces), security-sensitive details of buildings and infrastructure (for example, which doors
  have broken locks or weak points in a water system's controls), passwords and keys.

**Tool types**
- `public_ai` — any free or personal-account AI tool: chatbots, note-taking and transcription apps, image generators.
- `enterprise_ai` — the firm-licensed AI assistant with a data contract.
- `local_model` — an open model running only on a firm computer.

**Rules:** `"yes"` (allowed), `"no"` (not allowed), or `"ask"` (only with written approval from the project manager,
and only if the client contract allows).

A **dictionary of dictionaries** (Lab 01) stores the table: `policy["Confidential"]["local_model"]` is the rule for
confidential client information in a local model.

> 💡 **Hint:** Ask: if this leaked, who would be harmed, and does the tool's contract protect it? Text values need quotes: `"no"`. There is room for debate on `enterprise_ai` and `local_model` — be ready to defend your choice.

```python
# TODO: replace every None with "yes", "no", or "ask". Discuss each one as a team.
# Keys: "Confidential" = Confidential (client); "Restricted" = Restricted (personal & security-sensitive)
policy = {
    "Public":       {"public_ai": "yes", "enterprise_ai": "yes", "local_model": "yes"},
    "Internal":     {"public_ai": None,  "enterprise_ai": None,  "local_model": None},
    "Confidential": {"public_ai": None,  "enterprise_ai": None,  "local_model": None},
    "Restricted":   {"public_ai": None,  "enterprise_ai": None,  "local_model": None},
}
pd.DataFrame(policy).T
```

```python
# ✅ Self-check: run this cell to test your work above
allowed = {"yes", "no", "ask"}
for data_class, row in policy.items():
    for tool, rule in row.items():
        assert rule in allowed, f"policy['{data_class}']['{tool}'] is {rule!r} - use 'yes', 'no', or 'ask'."
assert policy["Confidential"]["public_ai"] == "no", "Client documents in a free public AI tool? Re-read Part 1."
assert policy["Restricted"]["public_ai"] == "no", "Personal or security-sensitive data never goes into a public AI tool."
print("Your table is complete. Be ready to defend every 'yes' and 'ask'.")
```

This small helper looks up your rule. You will use it in Part 3.

```python
def may_i_paste(data_class, tool):
    """Print the team's rule for putting `data_class` information into `tool`."""
    meaning = {"yes": "allowed.",
               "no": "NOT allowed.",
               "ask": "only with written approval from the project manager (and only if the client contract allows)."}
    rule = policy[data_class][tool]
    print(f"{data_class} information -> {tool}: {meaning.get(rule, rule)}")

may_i_paste("Public", "public_ai")
```
